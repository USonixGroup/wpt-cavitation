```python
sys.path.append("..")
from wavelet_transform.utils.save_plots import save_figure

from wavelet_transform.postprocess import plotting as plot
import wavelet_transform.utils.kmeans_results as results
```

```python
savefigs_flag = False
```

### Load data

```python
voltage_pcd_number = [(80, 1), (280, 1)]
n = 1  # number of experiments

frequency = 1.1e6

exp_results = list()
for id in voltage_pcd_number:
    # Find average and std across multiple experiments of the same id
    classifying_metrics = list()
    centers = list()
    label_maps = list()
    for i in range(n):
        result_id = f"{id[0]}_{id[1]}"
        if n > 1:
            result_id = f"{result_id}_{i}"  # append experiment nunmber if relevant
        classifying_metric, centre, label_map = results.get_data(result_id)
        classifying_metrics.append(classifying_metrics)
        centers.append(centre)
        label_maps.append(label_map)

    # Append tuple containing centres_avg, centres_std, unique_metrics, label_map
    # representing each set of experiments as a single average
    exp_results.append(plot.get_center_metrics(classifying_metrics, centers, label_maps))
```

```python
exp_labels = ["low intensity", "high intensity"]

classifying_metrics = list()
centers = list()
label_maps = list()
error_bars = list()
for centres_avg, centres_std, unique_metrics, label_map in exp_results:
    classifying_metrics.append(unique_metrics)
    centers.append(centres_avg)
    label_maps.append(label_map)
    error_bars.append(centres_std)

fig1 = plot.plot_multiple_centers(
    classifying_metrics, centers, label_maps, exp_labels, error_bars=error_bars
)

if savefigs_flag is True:
    save_figure(fig1, "results_error_bars")
```
