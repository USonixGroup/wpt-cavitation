```python
sys.path.append("..")
import h5py
import numpy as np

import plotly.graph_objects as go
```

```python
path = "/Users/max/Documents/coding/WPT-Cavitation/results_acd_2_april_2025/130mvpp_50v/RFData/RFdata_02-April-2025_18-12-31.mat"    
```

```python
f = h5py.File(path,'r') 
ref = f["RcvData"]
obj_ref = ref[0][0]
obj = f[obj_ref]
data = np.array(obj)
print(data.shape)
```

```python
channel_num = [i for i in range(64)]
frame_num = 50
sample_rate = 20e6
c = 1450

channel_data = data[frame_num,channel_num,:]
fig = go.Figure()
mmps = c / sample_rate * 1000 / 2

for i in range(len(channel_num)):
    single_channel_data = channel_data[i, :]
    fig.add_trace(
        go.Scatter(
            x = [idx * mmps for idx in range(len(single_channel_data))],
            y = single_channel_data,
            name = f"channel {i+1}"
        )
    )

fig.update_layout(
    xaxis_title = "distance",
    yaxis_title = "amplitude"
)
fig.show()
```
