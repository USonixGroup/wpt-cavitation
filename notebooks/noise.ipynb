```python
import numpy as np
from scipy.stats import kurtosis as spkurt
import plotly.graph_objects as go
import matplotlib.pyplot as plt
from scipy.signal.windows import gaussian
```

```python
np.random.seed = 10
results = []
results2 = []
lengths = [i for i in range(10, 500, 5)]
for length in lengths:
    noise = np.random.randn(length)
    transient = gaussian(len(noise), 1) * 5
    noise += transient
    results.append(kurtosis(noise))
    results2.append(spkurt(noise, fisher=False))

plt.plot(transient+noise)
fig = go.Figure()
fig.add_trace(
    go.Scatter(x=lengths, y=results)
)
fig.add_trace(
    go.Scatter(x=lengths, y=results2)
)

fig.show()
```
